```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('logs_SPEDIA.csv')
```

```python
data.head()
```

```python
data.shape
```

```python
data.drop(columns=['Url','Size','Size_before','Size_after','Filename','Path','Cc','Bcc'],inplace=True)
```

```python
# Count all unique values in the 'Description' column
print(data['Description'].value_counts())
```

```python
# Keep only rows with the specified Description values
filtered_data = data[data['Description'].isin([
    'Audit: Highly Suspicious Command executed: /bin/nc',
    'Audit: Lowly Suspicious Command executed: /usr/bin/head'
])]
filtered_data.reset_index(drop=True, inplace=True)
filtered_data.head()
```

```python
filtered_data.shape
```

```python
filtered_data['Description']=filtered_data['Description'].map({'Audit: Highly Suspicious Command executed: /bin/nc':0,'Audit: Lowly Suspicious Command executed: /usr/bin/head':1})
```

```python
filtered_data.head()
```

```python
filtered_data['Content'].isnull().sum()
```

```python
# Calculate the number of NaN values in 'Content', 'To', and 'From' columns
nan_content = filtered_data['Content'].isnull().sum()
nan_to = filtered_data['To'].isnull().sum()
nan_from = filtered_data['From'].isnull().sum()
print(f"NaN in Content: {nan_content}")
print(f"NaN in To: {nan_to}")
print(f"NaN in From: {nan_from}")
```

```python
filtered_data=filtered_data.drop(columns=['Level','Command'])
```
